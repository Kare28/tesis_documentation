### con la base ponderada

````r
# ==========================================
# 1. INSTALAR Y CARGAR PAQUETE
# ==========================================

install.packages("Hmisc")
library(Hmisc)

# ==========================================
# 2. VARIABLES NUMÉRICAS
# ==========================================

numericas <- c(
  "antiguedad_final",
  "focos",
  "focos_ahor",
  "tot_resid"
)

# ==========================================
# 3. ESTADÍSTICOS PONDERADOS
# ==========================================

estadisticos_ponderados <- do.call(
  rbind,
  lapply(numericas, function(variable) {

    x <- df[[variable]]
    w <- df$factor

    # Eliminar valores faltantes y pesos no válidos
    validos <- !is.na(x) & !is.na(w) & w > 0

    x <- x[validos]
    w <- w[validos]

    # Media ponderada
    media <- weighted.mean(x, w)

    # Cuartiles y mediana ponderados
    cuartiles <- Hmisc::wtd.quantile(
      x,
      weights = w,
      probs = c(0.25, 0.50, 0.75)
    )

    # Desviación estándar ponderada
    desv <- sqrt(
      sum(w * (x - media)^2) / sum(w)
    )

    # Tabla de resultados
    data.frame(
      Variable = variable,
      Minimo = min(x),
      Q1 = cuartiles[1],
      Mediana = cuartiles[2],
      Media = media,
      Desv_Est = desv,
      Q3 = cuartiles[3],
      Maximo = max(x)
    )
  })
)

# ==========================================
# 4. REDONDEAR RESULTADOS
# ==========================================

rownames(estadisticos_ponderados) <- NULL

estadisticos_ponderados[-1] <- lapply(
  estadisticos_ponderados[-1],
  round,
  digits = 2
)

# Mostrar tabla
estadisticos_ponderados
````

````r
# ==================================================
# 1. LIBRERÍAS
# ==================================================

library(ggplot2)
library(dplyr)
library(Hmisc)
library(scales)

# ==================================================
# 2. VARIABLES NUMÉRICAS
# ==================================================

numericas <- c(
  "antiguedad_final",
  "focos",
  "focos_ahor",
  "tot_resid"
)

etiquetas <- c(
  "Antigüedad de la vivienda (años)",
  "Número total de focos",
  "Número de focos ahorradores",
  "Número de residentes"
)

# ==================================================
# 3. CALCULAR ESTADÍSTICOS PONDERADOS PARA BOXPLOT
# ==================================================

resultados <- lapply(seq_along(numericas), function(i) {

  variable <- numericas[i]

  x <- df[[variable]]
  w <- df$factor

  # Eliminar registros faltantes y pesos no válidos
  validos <- is.finite(x) & is.finite(w) & w > 0

  x <- x[validos]
  w <- w[validos]

  # Cuartiles ponderados
  q <- Hmisc::wtd.quantile(
    x,
    weights = w,
    probs = c(0.25, 0.50, 0.75)
  )

  Q1 <- unname(q[1])
  Me <- unname(q[2])
  Q3 <- unname(q[3])

  # Rango intercuartílico ponderado
  RIQ <- Q3 - Q1

  # Límites para detectar posibles valores atípicos
  limite_inf <- Q1 - 1.5 * RIQ
  limite_sup <- Q3 + 1.5 * RIQ

  # Bigotes: valores observados dentro de los límites
  bigote_inf <- min(
    x[x >= limite_inf & x <= limite_sup]
  )

  bigote_sup <- max(
    x[x >= limite_inf & x <= limite_sup]
  )

  # Valores observados fuera de los límites
  valores_atipicos <- sort(unique(
    x[x < limite_inf | x > limite_sup]
  ))

  resumen <- data.frame(
    Variable = etiquetas[i],
    Q1 = Q1,
    Me = Me,
    Q3 = Q3,
    Bigote_inf = bigote_inf,
    Bigote_sup = bigote_sup
  )

  atipicos <- data.frame(
    Variable = etiquetas[i],
    Valor = valores_atipicos
  )

  list(
    resumen = resumen,
    atipicos = atipicos
  )
})

# ==================================================
# 4. UNIR RESULTADOS
# ==================================================

estad_box <- bind_rows(
  lapply(resultados, `[[`, "resumen")
)

datos_atipicos <- bind_rows(
  lapply(resultados, `[[`, "atipicos")
)

# Mantener el orden de las variables
estad_box$Variable <- factor(
  estad_box$Variable,
  levels = etiquetas
)

datos_atipicos$Variable <- factor(
  datos_atipicos$Variable,
  levels = etiquetas
)

# Comprobar cuartiles ponderados
print(estad_box)

# ==================================================
# 5. CONSTRUIR BOXPLOT PONDERADO
# ==================================================

grafico_box_ponderado <- ggplot(estad_box) +

  # Bigote izquierdo
  geom_segment(
    aes(
      x = Bigote_inf,
      xend = Q1,
      y = 1,
      yend = 1
    ),
    linewidth = 0.7,
    color = "black"
  ) +

  # Bigote derecho
  geom_segment(
    aes(
      x = Q3,
      xend = Bigote_sup,
      y = 1,
      yend = 1
    ),
    linewidth = 0.7,
    color = "black"
  ) +

  # Extremo del bigote izquierdo
  geom_segment(
    aes(
      x = Bigote_inf,
      xend = Bigote_inf,
      y = 0.93,
      yend = 1.07
    ),
    linewidth = 0.7
  ) +

  # Extremo del bigote derecho
  geom_segment(
    aes(
      x = Bigote_sup,
      xend = Bigote_sup,
      y = 0.93,
      yend = 1.07
    ),
    linewidth = 0.7
  ) +

  # Caja Q1-Q3
  geom_rect(
    aes(
      xmin = Q1,
      xmax = Q3,
      ymin = 0.85,
      ymax = 1.15
    ),
    fill = "#22BFE6",
    color = "black",
    linewidth = 0.8
  ) +

  # Línea de la mediana
  geom_segment(
    aes(
      x = Me,
      xend = Me,
      y = 0.85,
      yend = 1.15
    ),
    linewidth = 1.1,
    color = "black"
  ) +

  # Posibles valores atípicos
  geom_point(
    data = datos_atipicos,
    aes(x = Valor, y = 1),
    inherit.aes = FALSE,
    color = "black",
    size = 1.4,
    alpha = 0.5
  ) +

  # ==================================================
  # 6. ETIQUETAS DE CUARTILES
  # ==================================================

  # Q1
  geom_text(
    aes(
      x = Q1,
      y = 0.48,
      label = paste0("Q1 = ", round(Q1, 0))
    ),
    size = 3.8,
    fontface = "bold",
    hjust = 0.5
  ) +

  # Mediana
  geom_text(
    aes(
      x = Me,
      y = 1.40,
      label = paste0("Me = ", round(Me, 0))
    ),
    size = 3.8,
    fontface = "bold",
    hjust = 0.5
  ) +

  # Q3
  geom_text(
    aes(
      x = Q3,
      y = 0.67,
      label = paste0("Q3 = ", round(Q3, 0))
    ),
    size = 3.8,
    fontface = "bold",
    hjust = 0.5
  ) +

  # ==================================================
  # 7. ORGANIZACIÓN DE LOS PANELES
  # ==================================================

  facet_wrap(
    ~ Variable,
    scales = "free_x",
    ncol = 1
  ) +

  # Malla cada 5 unidades
  scale_x_continuous(
    breaks = scales::breaks_width(5),
    expand = expansion(mult = c(0.08, 0.08))
  ) +

  # Espacio vertical
  scale_y_continuous(
    limits = c(0.25, 1.75),
    breaks = NULL,
    expand = expansion(mult = 0)
  ) +

  # ==================================================
  # 8. TÍTULOS Y NOTA
  # ==================================================

  labs(
    x = "Valor observado",
    y = NULL,

    caption = paste0(
      "Nota: Q1 = primer cuartil; ",
      "Me = mediana; Q3 = tercer cuartil.\n"
    )
  ) +

  # ==================================================
  # 9. ESTILO
  # ==================================================

  theme_classic(base_size = 12) +

  theme(

    strip.text = element_text(
      size = 11,
      face = "bold",
      margin = margin(t = 7, b = 7)
    ),

    strip.background = element_rect(
      fill = "#F7F7F7",
      color = "grey80",
      linewidth = 0.4
    ),

    axis.text.y = element_blank(),
    axis.ticks.y = element_blank(),

    axis.text.x = element_text(
      size = 11
    ),

    # Cuadrícula gris muy tenue
    panel.grid.major.x = element_line(
      color = "grey92",
      linewidth = 0.3
    ),

    panel.grid.minor = element_blank(),
    panel.grid.major.y = element_blank(),

    panel.border = element_rect(
      color = "grey80",
      fill = NA,
      linewidth = 0.4
    ),

    panel.spacing = grid::unit(1.5, "lines"),

    plot.caption = element_text(
      size = 11,
      hjust = 0,
      color = "grey40",
      face = "plain",
      lineheight = 1.3,
      margin = margin(t = 15)
    ),

    plot.margin = margin(
      t = 15,
      r = 20,
      b = 15,
      l = 20
    )
  )

# ==================================================
# 10. MOSTRAR
# ==================================================

print(grafico_box_ponderado)

# ==================================================
# 11. GUARDAR PDF PARA LATEX
# ==================================================

ggsave(
  filename = "boxplot_variables_ponderadas.pdf",
  plot = grafico_box_ponderado,
  width = 10,
  height = 10.5,
  units = "in",
  device = "pdf"
)
````

Código en R para calcular las asociaciones
Este código utiliza los objetos que ya preparamos: df_asociaciones, variables_numericas, variables_ordinales, variables_nominales y variables_binarias.

````r

# =====================================================
# PASO 8. MATRIZ HETEROGÉNEA DE ASOCIACIONES
# ENIGH 2024 - COAHUILA
# =====================================================

# 1. IDENTIFICAR TIPO DE VARIABLE

obtener_tipo <- function(variable) {

  if (variable %in% variables_numericas) return("numerica")
  if (variable %in% variables_ordinales) return("ordinal")
  if (variable %in% variables_nominales) return("nominal")
  if (variable %in% variables_binarias) return("binaria")

  stop(paste("Variable no clasificada:", variable))
}

# 2. V DE CRAMÉR CORREGIDA POR SESGO

cramers_v_corregida <- function(x, y) {

  tabla <- table(x, y)
  tabla <- tabla[
    rowSums(tabla) > 0,
    colSums(tabla) > 0,
    drop = FALSE
  ]

  n <- sum(tabla)
  r <- nrow(tabla)
  k <- ncol(tabla)

  if (r < 2 || k < 2 || n <= 1) return(NA_real_)

  chi2 <- suppressWarnings(
    chisq.test(tabla, correct = FALSE)$statistic
  )

  phi2 <- as.numeric(chi2) / n

  # Corrección por sesgo
  phi2_corr <- max(
    0,
    phi2 - ((k - 1) * (r - 1)) / (n - 1)
  )

  r_corr <- r - ((r - 1)^2) / (n - 1)
  k_corr <- k - ((k - 1)^2) / (n - 1)

  denominador <- min(r_corr - 1, k_corr - 1)

  if (denominador <= 0) return(NA_real_)

  sqrt(phi2_corr / denominador)
}

# 3. COEFICIENTE PHI

coef_phi <- function(x, y) {

  # La codificación es No = 0, Sí = 1
  x_num <- as.integer(as.character(x) == "Sí")
  y_num <- as.integer(as.character(y) == "Sí")

  if (sd(x_num) == 0 || sd(y_num) == 0) {
    return(NA_real_)
  }

  cor(x_num, y_num, method = "pearson")
}

# 4. RAZÓN DE CORRELACIÓN ETA

coef_eta <- function(numerica, categorica) {

  grupos <- as.factor(categorica)

  if (nlevels(droplevels(grupos)) < 2) {
    return(NA_real_)
  }

  media_general <- mean(numerica)

  ss_total <- sum((numerica - media_general)^2)

  if (ss_total == 0) return(NA_real_)

  medias <- tapply(numerica, grupos, mean)
  frecuencias <- table(grupos)

  ss_entre <- sum(
    frecuencias[names(medias)] *
      (medias - media_general)^2
  )

  sqrt(max(0, min(1, ss_entre / ss_total)))
}

# 5. FUNCIÓN GENERAL DE ASOCIACIÓN

calcular_asociacion <- function(x, y, tipo_x, tipo_y) {

  tipos_ordenados <- c("numerica", "ordinal")

  # Numérica u ordinal contra numérica u ordinal
  if (tipo_x %in% tipos_ordenados &&
      tipo_y %in% tipos_ordenados) {

    valor <- suppressWarnings(
      cor(
        as.numeric(x),
        as.numeric(y),
        method = "spearman"
      )
    )

    return(list(valor = valor, metodo = "Spearman"))
  }

  # Binaria contra binaria
  if (tipo_x == "binaria" && tipo_y == "binaria") {

    return(list(
      valor = coef_phi(x, y),
      metodo = "Phi"
    ))
  }

  # Numérica contra nominal o binaria
  if (tipo_x == "numerica" &&
      tipo_y %in% c("nominal", "binaria")) {

    return(list(
      valor = coef_eta(x, y),
      metodo = "Eta"
    ))
  }

  if (tipo_y == "numerica" &&
      tipo_x %in% c("nominal", "binaria")) {

    return(list(
      valor = coef_eta(y, x),
      metodo = "Eta"
    ))
  }

  # Resto de comparaciones categóricas
  return(list(
    valor = cramers_v_corregida(x, y),
    metodo = "V de Cramer corregida"
  ))
}

# 6. CREAR MATRICES

nombres <- names(df_asociaciones)
n <- length(nombres)

matriz_asociaciones <- matrix(
  NA_real_,
  nrow = n,
  ncol = n,
  dimnames = list(nombres, nombres)
)

matriz_metodos <- matrix(
  "",
  nrow = n,
  ncol = n,
  dimnames = list(nombres, nombres)
)

# 7. CALCULAR CADA PAREJA UNA SOLA VEZ

for (i in seq_len(n)) {

  for (j in i:n) {

    if (i == j) {
      matriz_asociaciones[i, j] <- 1
      matriz_metodos[i, j] <- "Diagonal"
      next
    }

    resultado <- calcular_asociacion(
      df_asociaciones[[i]],
      df_asociaciones[[j]],
      obtener_tipo(nombres[i]),
      obtener_tipo(nombres[j])
    )

    matriz_asociaciones[i, j] <- resultado$valor
    matriz_asociaciones[j, i] <- resultado$valor

    matriz_metodos[i, j] <- resultado$metodo
    matriz_metodos[j, i] <- resultado$metodo
  }
}

# 8. VERIFICACIÓN

cat("Dimensión de la matriz:\n")
print(dim(matriz_asociaciones))

cat("\nMétodos utilizados:\n")
print(table(matriz_metodos[upper.tri(matriz_metodos)]))

cat("\nValores NA:\n")
print(sum(is.na(matriz_asociaciones)))

cat("\nVista preliminar:\n")
print(round(matriz_asociaciones[1:6, 1:6], 3))
````

````r
library(dplyr)
library(tidyr)

# ==========================================
# 1. ORDEN DE LOS ESTRATOS
# ==========================================

orden_estratos <- c(
  "Bajo",
  "Medio bajo",
  "Medio alto",
  "Alto"
)

# ==========================================
# 2. FUNCIÓN PARA TABLAS CRUZADAS
# ==========================================

crear_tabla_estrato <- function(variable) {

  tabla <- df_asociaciones %>%
    transmute(
      estrato = factor(
        as.character(est_socio_desc),
        levels = orden_estratos
      ),
      material = as.character(.data[[variable]])
    ) %>%
    filter(
      !is.na(estrato),
      !is.na(material)
    ) %>%
    count(
      estrato,
      material,
      name = "frecuencia"
    ) %>%
    complete(
      estrato,
      material,
      fill = list(frecuencia = 0)
    ) %>%
    group_by(estrato) %>%
    mutate(
      total_estrato = sum(frecuencia),
      porcentaje = round(
        frecuencia / total_estrato * 100,
        2
      )
    ) %>%
    ungroup() %>%
    arrange(estrato, material)

  return(tabla)
}

# ==========================================
# 3. ESTRATO × MATERIAL DE TECHOS
# ==========================================

tabla_estrato_techos <- crear_tabla_estrato(
  "techos_desc"
)

print(tabla_estrato_techos, n = Inf)

# ==========================================
# 4. ESTRATO × MATERIAL DE PAREDES
# ==========================================

tabla_estrato_paredes <- crear_tabla_estrato(
  "mat_pared_desc"
)

print(tabla_estrato_paredes, n = Inf)
````

````r
# ==========================================
# FUNCIÓN PARA TABLA RESUMIDA
# ==========================================

crear_tabla_resumen <- function(tabla) {

  tabla %>%
    mutate(
      resultado = paste0(
        frecuencia,
        " (",
        sprintf("%.2f", porcentaje),
        " %)"
      )
    ) %>%
    select(
      estrato,
      material,
      resultado
    ) %>%
    pivot_wider(
      names_from = material,
      values_from = resultado
    )
}

# Tabla resumida de techos
resumen_estrato_techos <- crear_tabla_resumen(
  tabla_estrato_techos
)

# Tabla resumida de paredes
resumen_estrato_paredes <- crear_tabla_resumen(
  tabla_estrato_paredes
)

# Mostrar resultados completos
print(resumen_estrato_techos, width = Inf)
print(resumen_estrato_paredes, width = Inf)
````

````r
crear_tabla <- function(data, variable, factor_exp = "factor") {

  tabla <- data %>%
    group_by(.data[[variable]]) %>%
    summarise(
      cantidad_muestra = n(),
      viviendas_reales = sum(.data[[factor_exp]], na.rm = TRUE)
    ) %>%
    ungroup() %>%
    mutate(
      pct_muestra = round(cantidad_muestra / sum(cantidad_muestra) * 100, 2),
      pct_reales = round(viviendas_reales / sum(viviendas_reales) * 100, 2)
    ) %>%
    arrange(desc(viviendas_reales))

  return(tabla)
}
````

#GRAFICOS DE ELECTRODOMOESTICOS

En relación con el equipamiento destinado a la preparación y conservación de alimentos, la Tabla 3.X presenta las frecuencias muestrales y expandidas correspondientes a las viviendas que disponen de estufa, refrigerador, licuadora y horno de microondas. De acuerdo con las estimaciones obtenidas mediante los factores de expansión de la ENIGH 2024, se observa una mayor disponibilidad de estufa y refrigerador, mientras que el horno de microondas presenta una menor proporción de viviendas equipadas.
Figura 3.X. Distribución porcentual muestral del equipamiento relacionado con la cocina en Coahuila, ENIGH 2024.

Tabla 3.X. Frecuencias muestrales y expandidas de las viviendas de Coahuila según la disponibilidad de equipamiento de cocina, ENIGH 2024.

````r
library(dplyr)
library(tidyr)
library(ggplot2)
library(scales)

# ===============================
# 1. Definir grupos de variables binarias
# ===============================

grupos_binarias <- list(
  "Aseo personal" = c(
    "calentsol_desc", "calentgas_desc", "calentlena_desc", "ten_lavad_desc"
  ),
  "Cocina" = c(
    "ten_refri_desc", "ten_estuf_desc", "ten_licua_desc", "ten_micro_desc"
  ),
  "Confort térmico" = c(
    "ten_venti_desc", "aireacond_desc", "calefacc_desc"
  ),
  "Comunicación" = c(
    "ten_tva_desc", "ten_tvd_desc", "ten_radio_desc",
    "ten_lap_desc", "ten_compu_desc"
  )
)

# ===============================
# 2. Función para gráfico divergente
# ===============================

grafico_divergente <- function(data, variables, titulo) {

  tabla <- data %>%
    select(all_of(variables)) %>%
    pivot_longer(
      cols = everything(),
      names_to = "variable",
      values_to = "respuesta"
    ) %>%
    count(variable, respuesta) %>%
    group_by(variable) %>%
    mutate(
      porcentaje = n / sum(n) * 100,
      porcentaje_div = ifelse(respuesta == "No", -porcentaje, porcentaje),
      etiqueta = paste0(round(porcentaje, 1), "%")
    ) %>%
    ungroup()

  ggplot(
    tabla,
    aes(
      x = porcentaje_div,
      y = reorder(variable, porcentaje_div),
      fill = respuesta
    )
  ) +
    geom_col(width = 0.65) +
    geom_text(
      aes(
        label = etiqueta,
        hjust = ifelse(porcentaje_div < 0, 1.1, -0.1)
      ),
      size = 4,
      fontface = "bold"
    ) +
    scale_x_continuous(
      labels = function(x) paste0(abs(x), "%"),
      limits = c(-100, 100),
      breaks = seq(-100, 100, 25)
    ) +
    scale_fill_manual(
      values = c(
        "No" = "#D95F5F",
        "Sí" = "#2A9D8F"
      )
    ) +
    labs(
      title = titulo,
      subtitle = "Distribución porcentual de respuestas Sí y No",
      x = "Porcentaje de viviendas",
      y = NULL,
      fill = "Respuesta",
      caption = "Fuente: elaboración propia con base en ENIGH 2024"
    ) +
    theme_minimal() +
    theme(
      plot.title = element_text(face = "bold", size = 16, hjust = 0.5),
      plot.subtitle = element_text(size = 11, hjust = 0.5),
      axis.text.y = element_text(size = 11),
      panel.grid.major.y = element_blank(),
      panel.grid.minor = element_blank(),
      legend.position = "bottom"
    )
}

# ===============================
# 3. Generar gráficos
# ===============================

grafico_divergente(
  df,
  grupos_binarias[["Aseo personal"]],
  "Equipamiento relacionado con aseo personal"
)

grafico_divergente(
  df,
  grupos_binarias[["Cocina"]],
  "Equipamiento relacionado con cocina"
)

grafico_divergente(
  df,
  grupos_binarias[["Confort térmico"]],
  "Equipamiento relacionado con confort térmico"
)

grafico_divergente(
  df,
  grupos_binarias[["Comunicación"]],
  "Equipamiento relacionado con comunicación"
)
````

````r
library(dplyr)
library(ggplot2)

# =========================================================
# 1. ORDEN DE VARIABLES SEGÚN TUS BLOQUES CONCEPTUALES
# =========================================================

orden_variables <- unlist(bloques, use.names = FALSE)

# Reordenar la matriz
matriz_ordenada <- matriz_asociaciones[orden_variables, orden_variables]

# =========================================================
# 2. ETIQUETAS DESCRIPTIVAS
# =========================================================
# Ajusté los nombres técnicos a los exactos de tu base

etiquetas <- c(
  "mat_pared_desc"      = "Mat. pared",
  "techos_desc"         = "Mat. techos",
  "pisos_desc"          = "Mat. pisos",
  "tipoviv_desc"        = "Tipo viv.",
  "antiguedad_final"    = "Antigüedad",
  "tot_resid"           = "Residentes",
  "cocina_desc"         = "Cocina",
  "dispElect_desc"      = "Energía eléctr.",
  "pelectric_desc"      = "Problema eléctr.",
  "region_bioclimatica" = "Bioclima",
  "calefacc_desc"       = "Calefacción",
  "aireacond_desc"      = "Aire acond.",
  "ten_venti_desc"      = "Ventilador",
  "calentsol_desc"      = "Boiler solar",
  "calentgas_desc"      = "Boiler gas",
  "calentlena_desc"     = "Boiler leña",
  "ten_micro_desc"      = "Microondas",
  "ten_licua_desc"      = "Licuadora",
  "ten_estuf_desc"      = "Estufa",
  "ten_refri_desc"      = "Refri",
  "ten_lavad_desc"      = "Lavadora",
  "ten_compu_desc"      = "Computadora",
  "ten_lap_desc"        = "Laptop",
  "ten_tva_desc"        = "TVA",
  "ten_tvd_desc"        = "TVD",
  "ten_radio_desc"      = "Radio",
  "focos"               = "Focos",
  "focos_ahor"          = "Focos ahor.",
  "ilumina_ef"          = "Ilumina. ef.",
  "est_socio_desc"      = "Estrato socioec."
)

# =========================================================
# 3. SEPARADORES ENTRE BLOQUES
# =========================================================

n <- length(orden_variables)

tam_bloques <- lengths(bloques)
cortes <- cumsum(tam_bloques)
cortes <- cortes[-length(cortes)]

separadores_x <- cortes + 0.5
separadores_y <- n - cortes + 0.5

# =========================================================
# 4. FUNCIÓN PARA CREAR EL MAPA DE CALOR
# =========================================================

crear_mapa_asociaciones <- function(matriz,
                                    etiquetas,
                                    orden_variables,
                                    separadores_x,
                                    separadores_y,
                                    quitar_diagonal = FALSE,
                                    titulo = "Matriz heterogénea de asociaciones",
                                    subtitulo = "Coeficientes de asociación entre variables") {

  datos <- as.data.frame(as.table(matriz))
  names(datos) <- c("Variable_1", "Variable_2", "Coeficiente")

  datos <- datos %>%
    mutate(
      Variable_1 = as.character(Variable_1),
      Variable_2 = as.character(Variable_2),
      diagonal = Variable_1 == Variable_2,
      Intensidad = abs(Coeficiente),
      Etiqueta = sprintf("%.2f", Coeficiente)
    )

  # Si se quiere quitar la diagonal
  if (quitar_diagonal) {
    datos <- datos %>%
      mutate(
        Intensidad = ifelse(diagonal, NA, Intensidad),
        Etiqueta = ifelse(diagonal, "", Etiqueta)
      )
  }

  # Reordenar factores
  datos <- datos %>%
    mutate(
      Variable_1 = factor(Variable_1, levels = orden_variables),
      Variable_2 = factor(Variable_2, levels = rev(orden_variables))
    )

  grafico <- ggplot(
    datos,
    aes(
      x = Variable_1,
      y = Variable_2,
      fill = Intensidad
    )
  ) +
    geom_tile(color = "white", linewidth = 0.15) +

    geom_text(
      aes(
        label = Etiqueta,
        color = Intensidad > 0.60
      ),
      size = 2.2
    ) +

    scale_color_manual(
      values = c(
        "FALSE" = "#243746",
        "TRUE"  = "white"
      ),
      guide = "none",
      na.value = "#243746"
    ) +

    geom_vline(
      xintercept = separadores_x,
      color = "grey30",
      linewidth = 0.45
    ) +

    geom_hline(
      yintercept = separadores_y,
      color = "grey30",
      linewidth = 0.45
    ) +

    scale_fill_gradient(
      low = "#F2F7FA",
      high = "#1479A8",
      limits = c(0, 1),
      na.value = "white",
      name = "|Asociación|"
    ) +

    scale_x_discrete(labels = etiquetas) +
    scale_y_discrete(labels = etiquetas) +

    coord_fixed() +

    labs(
      title = titulo,
      subtitle = subtitulo,
      x = NULL,
      y = NULL,
      caption = paste(
        "ENIGH 2024, Coahuila (n = 4,090).",
        "Medidas: Spearman, Phi, V de Cramér corregida y Eta.",
        "El color representa la magnitud absoluta de la asociación."
      )
    ) +

    theme_minimal(base_size = 11) +
    theme(
      axis.text.x = element_text(
        angle = 90,
        hjust = 1,
        vjust = 0.5,
        size = 8
      ),
      axis.text.y = element_text(size = 8),
      panel.grid = element_blank(),
      plot.title = element_text(face = "bold", hjust = 0.5, size = 14),
      plot.subtitle = element_text(hjust = 0.5, size = 11),
      plot.caption = element_text(size = 8.5, hjust = 0.5),
      legend.position = "right"
    )

  return(grafico)
}

# =========================================================
# 5. GRÁFICO 1: MATRIZ COMPLETA CON DIAGONAL
# =========================================================

grafico_matriz_completa <- crear_mapa_asociaciones(
  matriz = matriz_ordenada,
  etiquetas = etiquetas,
  orden_variables = orden_variables,
  separadores_x = separadores_x,
  separadores_y = separadores_y,
  quitar_diagonal = FALSE,
  titulo = "Matriz heterogénea de asociaciones",
  subtitulo = "Matriz completa con diagonal y coeficientes numéricos"
)

print(grafico_matriz_completa)

# Guardar
ggsave(
  filename = "matriz_asociaciones_completa_etiquetas.pdf",
  plot = grafico_matriz_completa,
  width = 17,
  height = 14,
  units = "in"
)

````

En general, el análisis permitió identificar relaciones entre variables que describen distintas condiciones habitacionales y energéticas, sin que ello implique necesariamente causalidad o justifique su exclusión automática del conjunto de datos.

````r

# =====================================================
# MATRIZ HETEROGÉNEA DE ASOCIACIONES
# ENIGH 2024 - COAHUILA
# =====================================================

# Instalar solo si hace falta:
 install.packages("ggnewscale")

library(dplyr)
library(ggplot2)
library(ggnewscale)

# =====================================================
# 1. ORDEN DE VARIABLES POR BLOQUES
# =====================================================

orden_variables <- unlist(
  bloques,
  use.names = FALSE
)

n_vars <- length(orden_variables)

# Reordenar matriz
matriz_ordenada <- matriz_asociaciones[
  orden_variables,
  orden_variables
]


# =====================================================
# 2. ETIQUETAS DESCRIPTIVAS
# =====================================================

etiquetas <- c(
  "mat_pared_desc"      = "Mat. pared",
  "techos_desc"         = "Mat. techos",
  "pisos_desc"          = "Mat. pisos",
  "tipoviv_desc"        = "Tipo viv.",
  "antiguedad_final"    = "Antigüedad",
  "tot_resid"           = "Residentes",
  "cocina_desc"         = "Cocina",
  "dispElect_desc"      = "Energía eléctr.",
  "pelectric_desc"      = "Problema eléctr.",
  "region_bioclimatica" = "Bioclima",
  "calefacc_desc"       = "Calefacción",
  "aireacond_desc"      = "Aire acond.",
  "ten_venti_desc"      = "Ventilador",
  "calentsol_desc"      = "Boiler solar",
  "calentgas_desc"      = "Boiler gas",
  "calentlena_desc"     = "Boiler leña",
  "ten_micro_desc"      = "Microondas",
  "ten_licua_desc"      = "Licuadora",
  "ten_estuf_desc"      = "Estufa",
  "ten_refri_desc"      = "Refri",
  "ten_lavad_desc"      = "Lavadora",
  "ten_compu_desc"      = "Computadora",
  "ten_lap_desc"        = "Laptop",
  "ten_tva_desc"        = "TVA",
  "ten_tvd_desc"        = "TVD",
  "ten_radio_desc"      = "Radio",
  "focos"               = "Focos",
  "focos_ahor"          = "Focos ahor.",
  "ilumina_ef"          = "Ilumina. ef.",
  "est_socio_desc"      = "Estrato socioec."
)


# =====================================================
# 3. COLORES DE LAS DIMENSIONES
# =====================================================

colores_bloques <- c(
  "Constructivas y habitacionales" = "#D9A15B",
  "Acceso a fuentes de energia" = "#7E9FD9",
  "Confort termico y calentamiento de agua" = "#5AB1A7",
  "Alimentacion e higiene" = "#4C9F70",
  "Comunicacion y entretenimiento" = "#8A6CCF",
  "Iluminacion" = "#3B82C4",
  "Condicion socioeconomica" = "#C86D9E"
)


# =====================================================
# 4. MATRIZ A FORMATO LARGO
# =====================================================

datos_heatmap <- as.data.frame(
  as.table(matriz_ordenada)
)

colnames(datos_heatmap) <- c(
  "Variable_1",
  "Variable_2",
  "Asociacion"
)

datos_heatmap <- datos_heatmap %>%
  mutate(
    Intensidad = abs(Asociacion),
    x = match(Variable_1, orden_variables),
    y = match(
      Variable_2,
      rev(orden_variables)
    )
  )


# =====================================================
# 5. VARIABLES Y BLOQUES PARA ANOTACIONES
# =====================================================

bloque_por_variable <- rep(
  names(bloques),
  lengths(bloques)
)

# Franja superior
anotacion_superior <- data.frame(
  variable = orden_variables,
  bloque = bloque_por_variable,
  x = 1:n_vars,
  y = n_vars + 0.8
)

# Franja lateral izquierda
anotacion_lateral <- data.frame(
  variable = rev(orden_variables),
  bloque = rev(bloque_por_variable),
  x = 0.2,
  y = 1:n_vars
)


# =====================================================
# 6. SEPARADORES ENTRE BLOQUES
# =====================================================

tam_bloques <- lengths(bloques)

separadores <- cumsum(tam_bloques)

separadores <- separadores[
  -length(separadores)
] + 0.5


# =====================================================
# 7. CONSTRUCCIÓN DE LA MATRIZ
# =====================================================

grafico_asociaciones_bloques <- ggplot() +

  # -------------------------------------------------
  # A. MAPA DE CALOR
  # -------------------------------------------------

  geom_tile(
    data = datos_heatmap,
    aes(
      x = x,
      y = y,
      fill = Intensidad
    ),
    color = "white",
    linewidth = 0.15
  ) +

  # Escala de asociación
  scale_fill_gradient(
    low = "#F2F7FA",
    high = "#1479A8",
    limits = c(0, 1),
    breaks = seq(0, 1, 0.25),
    name = "|Asociación|",

    guide = guide_colorbar(
      direction = "horizontal",
      barwidth = grid::unit(9, "cm"),
      barheight = grid::unit(0.35, "cm"),
      title.position = "top",
      order = 1
    )
  ) +

  # -------------------------------------------------
  # B. LÍNEAS QUE SEPARAN LOS BLOQUES
  # -------------------------------------------------

  geom_vline(
    xintercept = separadores,
    color = "grey30",
    linewidth = 0.45
  ) +

  geom_hline(
    yintercept = n_vars - separadores + 1,
    color = "grey30",
    linewidth = 0.45
  ) +

  # -------------------------------------------------
  # C. NUEVA ESCALA PARA LAS DIMENSIONES
  # -------------------------------------------------

  ggnewscale::new_scale_fill() +

  # Franja superior
  geom_tile(
    data = anotacion_superior,
    aes(
      x = x,
      y = y,
      fill = bloque
    ),
    height = 0.35
  ) +

  # Franja lateral
  geom_tile(
    data = anotacion_lateral,
    aes(
      x = x,
      y = y,
      fill = bloque
    ),
    width = 0.35
  ) +

  # Colores y leyenda de dimensiones
  scale_fill_manual(
    values = colores_bloques,
    name = "Dimensión",

    guide = guide_legend(
      ncol = 3,
      byrow = TRUE,
      title.position = "top",
      order = 2
    )
  ) +

  # -------------------------------------------------
  # D. ESCALAS Y ETIQUETAS DE VARIABLES
  # -------------------------------------------------

  scale_x_continuous(
    breaks = 1:n_vars,
    labels = etiquetas[orden_variables],
    expand = c(0, 0)
  ) +

  scale_y_continuous(
    breaks = 1:n_vars,
    labels = etiquetas[rev(orden_variables)],
    expand = c(0, 0)
  ) +

  coord_fixed(clip = "off") +

  # -------------------------------------------------
  # E. TÍTULOS Y NOTA
  # -------------------------------------------------

  labs(
    #title = "Matriz heterogénea de asociaciones",


    #subtitle = "Viviendas de Coahuila, ENIGH 2024",

    x = NULL,
    y = NULL,


  ) +

  # =================================================
  # 8. APARIENCIA GENERAL DEL GRÁFICO
  # AQUÍ PUEDES MODIFICAR LOS TAMAÑOS
  # =================================================

  theme_minimal(base_size = 11) +

  theme(

    # -----------------------------------------------
    # A. VARIABLES DEL EJE HORIZONTAL (X)
    # -----------------------------------------------

    axis.text.x = element_text(
      angle = 90,
      hjust = 1,
      vjust = 0.5,
      size = 11,          # Tamaño de letra
      #face = "bold",      # Negritas
      color = "black"
    ),

    # -----------------------------------------------
    # B. VARIABLES DEL EJE VERTICAL (Y)
    # -----------------------------------------------

    axis.text.y = element_text(
      size = 11,          # Tamaño de letra
      #face = "bold",      # Negritas
      color = "black"
    ),

    # -----------------------------------------------
    # C. CUADRÍCULA INTERNA
    # -----------------------------------------------

    panel.grid = element_blank(),

    # -----------------------------------------------
    # D. TÍTULO
    # -----------------------------------------------

    plot.title = element_text(
      face = "bold",
      hjust = 0.5,
      size = 14
    ),

    # -----------------------------------------------
    # E. SUBTÍTULO
    # -----------------------------------------------

    plot.subtitle = element_text(
      hjust = 0.5,
      size = 12
    ),


    # -----------------------------------------------
    # G. POSICIÓN DE LAS LEYENDAS
    # -----------------------------------------------

    legend.position = "bottom",

    legend.box = "vertical",

    legend.box.just = "center",

    # -----------------------------------------------
    # H. TÍTULOS DE LAS LEYENDAS
    # -----------------------------------------------

    legend.title = element_text(
      size = 12,
      face = "bold"
    ),

    # -----------------------------------------------
    # I. TEXTO DE LAS LEYENDAS
    # -----------------------------------------------

    legend.text = element_text(
      size = 12
    ),

    # -----------------------------------------------
    # J. SEPARACIÓN ENTRE LEYENDAS
    # -----------------------------------------------

    legend.spacing.y = grid::unit(0.4, "cm"),

    # -----------------------------------------------
    # K. MÁRGENES DE LAS LEYENDAS
    # -----------------------------------------------

    legend.margin = margin(
      t = 5,
      r = 5,
      b = 5,
      l = 5
    ),

    # -----------------------------------------------
    # L. MÁRGENES GENERALES
    # -----------------------------------------------

    plot.margin = margin(
      t = 20,
      r = 25,
      b = 20,
      l = 40
    )
  )


# =====================================================
# 9. MOSTRAR GRÁFICO
# =====================================================

print(grafico_asociaciones_bloques)


# =====================================================
# 10. GUARDAR EN PDF
# =====================================================

ggsave(
  filename = "matriz_asociaciones_bloques_leyenda.pdf",
  plot = grafico_asociaciones_bloques,
  width = 11,
  height = 11,
  units = "in",
  device = "pdf"
)
````
